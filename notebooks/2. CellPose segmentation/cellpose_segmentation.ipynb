```python
import tifffile as tiff
from cellpose import core, io, models
from tqdm import tqdm

io.logger_setup() # run this to get printing of progress

#Check if colab notebook instance has GPU access
if core.use_gpu()==False:
  raise ImportError("No GPU access, change my runtime")

model = models.CellposeModel(gpu=True)
```

```python
flow_threshold = 0.4
cellprob_threshold = 0.0
tile_norm_blocksize = 0
```

```python
folder_path = "/path/to/my/folder"  
bf_images = [file for file in os.listdir(folder_path) if ("Bright" in file) and ("mask" not in file)]
masks = [file for file in os.listdir(folder_path) if "mask" in file]
print(len(bf_images))
print(len(masks))
```

```python
for bf_image in tqdm(bf_images, desc="Processing images"):
  mask_file = os.path.join(folder_path, bf_image.replace(".tif", "_mask.tif"))
  if mask_file in masks:
    continue
  else:
    # print(mask_file)
    img = io.imread(os.path.join(folder_path, bf_image))
    mask, flow, style = model.eval(
        img,
        batch_size=32,
        flow_threshold=flow_threshold,
        cellprob_threshold=cellprob_threshold,
        normalize={"tile_norm_blocksize": tile_norm_blocksize})
    tiff.imwrite(mask_file, mask)
```
