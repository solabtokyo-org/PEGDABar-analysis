```python
import os
```

```python
def get_well_name_fields_dict(cloud_path):
    file_names = [file for file in os.listdir(cloud_path) if file.endswith(".tif")]
    well_names = sorted(set([file_name.split("(")[0] for file_name in file_names]))
    well_name_fields_dict = {}
    for well_name in well_names:
        file_names_for_well = [file_name for file_name in file_names if file_name.startswith(well_name)]
        field = sorted(set([int(file_name.split("(")[1].split(" ")[1]) for file_name in file_names_for_well ]))
        well_name_fields_dict[well_name] = field
    return well_name_fields_dict

def get_focused_dict(folder_path: str) -> dict:
    subfolders = [
        f.path for f in os.scandir(folder_path) if f.is_dir()
    ]

    focused_dict = {}
    for subfolder in subfolders:
        field_name = subfolder.split("/")[-1]
        well_name = field_name.split("_")[0]
        field_number = int(field_name.split("-")[-1])
        if well_name not in focused_dict:
            focused_dict[well_name] = []
        n_contents = len(os.listdir(subfolder))
        if n_contents > 0:
            focused_dict[well_name].append(field_number)
    return focused_dict

def get_n_focused_blurred(well_name_fields_dict: dict, focused_dict: dict):
    n_blurred_dict = {}
    n_focused_dict = {}
    for well_name in well_name_fields_dict.keys():
        all_fields = set(well_name_fields_dict[well_name])
        focused_fields = set(focused_dict.get(well_name, []))
        blurred_fields = all_fields - focused_fields
        n_blurred_dict[well_name] = len(blurred_fields)
        n_focused_dict[well_name] = len(focused_fields)
    return n_blurred_dict, n_focused_dict
```

```python
cloud_path_0529 = "/path/to/my/cloud/folder"  
well_name_fields_dict_0529 = get_well_name_fields_dict(cloud_path_0529)
focused_dict_0529 = get_focused_dict("/path/to/my/focused/folder")
n_blurred_dict, n_focused_dict = get_n_focused_blurred(well_name_fields_dict_0529, focused_dict_0529)
print("Date: 0529")
print("focused: ", n_focused_dict)
print("defocused: ", n_blurred_dict)
print("sum focused: ", sum(n_focused_dict.values()))
print("sum defocused: ", sum(n_blurred_dict.values()))
```
