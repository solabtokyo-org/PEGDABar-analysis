```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from scipy.optimize import minimize
from sklearn.mixture import GaussianMixture
```

```python
df_0529 = pd.read_csv("output/0529.csv")
df_0610 = pd.read_csv("output/0610.csv")
```

```python
df_0529['hash_simple'] = df_0529['hash_simple'].astype(str).str.zfill(5)
df_0610['hash_simple'] = df_0610['hash_simple'].astype(str).str.zfill(5)
```

```python
# Filter rows where 'hash_simple' as a string does not contain '0'
df_0529_no_zero = df_0529[df_0529['hash_simple'].astype(
    str).str.contains('0') == False]
df_0610_no_zero = df_0610[df_0610['hash_simple'].astype(
    str).str.contains('0') == False]
```

```python
hash_0529 = df_0529_no_zero["hash_simple"][::5]
hash_0610 = df_0610_no_zero["hash_simple"][::5]
```

```python
# Convert hash_0610 to a set (for faster lookup)
hash_0610_set = set(str(h).zfill(5) for h in hash_0610)

# Dictionary to count differences at each digit position
digit_diff_counts = {i: 0 for i in range(5)}

# Compare and extract hashes differing by one digit
for h1 in hash_0529:
    h1_str = str(h1).zfill(5)

    # Modify each digit and check for matches
    for pos in range(5):
        for digit in '123456789':
            if digit != h1_str[pos]:
                modified = h1_str[:pos] + digit + h1_str[pos+1:]
                if modified in hash_0610_set:
                    digit_diff_counts[pos] += 1

# Display the results
print("Number of one-digit differences at each position:")
for digit, count in digit_diff_counts.items():
    print(f"Layer {digit + 1}: {count}")
```

```python
hash_counts = hash_0529.value_counts()
```

```python
hash_counts = hash_0529.value_counts()

unique_count = (hash_counts == 1).sum()
count_twice = (hash_counts == 2).sum()
count_thrice = (hash_counts == 3).sum()
count_four = (hash_counts >= 4).sum()

count_sum = (unique_count + 3*count_thrice + 2*count_twice + 4*count_four)
undefined = len(hash_0529) - count_sum

unassigned = len(df_0529)//5 - len(hash_0529)

print("0529")
print(f"Unique values: {unique_count}")
print(f"Values appearing twice: {count_twice}")
print(f"Values appearing three times: {count_thrice}")
print(f"Values appearing more than four times: {count_four}")
print(f"Total counted values: {count_sum}")
print(f"Undefined values: {undefined}")
print(f"unassigned values: {unassigned}")
print("---------------------")

hash_counts = hash_0610.value_counts()

unique_count = (hash_counts == 1).sum()
count_twice = (hash_counts == 2).sum()
count_thrice = (hash_counts == 3).sum()
count_four = (hash_counts >= 4).sum()
count_sum = (unique_count + 3*count_thrice + 2*count_twice + 4*count_four)
undefined = len(hash_0610) - count_sum
unassigned = len(df_0610)//5 - len(hash_0610)

print("0610")
print(f"Unique values: {unique_count}")
print(f"Values appearing twice: {count_twice}")
print(f"Values appearing three times: {count_thrice}")
print(f"Values appearing more than four times: {count_four}")
print(f"Total counted values: {count_sum}")
print(f"Undefined values: {undefined}")
print(f"unassigned values: {unassigned}")
print("---------------------")

hash_0529_unique_once = hash_0529.value_counts(
)[hash_0529.value_counts() == 1].index
hash_0610_unique_once = hash_0610.value_counts(
)[hash_0610.value_counts() == 1].index

common_hashes_once = np.intersect1d(
    hash_0529_unique_once, hash_0610_unique_once)
common_hashes_once
print(
    f"Intersection of counted onece between unique 0529 and 0610: {len(common_hashes_once)}")
```
