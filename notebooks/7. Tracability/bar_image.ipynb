```python
import tifffile as tiff
import numpy as np
import matplotlib.pyplot as plt
```

```python
original_image = tiff.imread(
    'images/0529/A - 3_fld-182/105/aligned_image.tif')
original_image_8bit = (
    original_image / np.max(original_image) * 255).astype(np.uint8)
c3_log_image = np.log(original_image[:, :, 2] + 1)
label_map = tiff.imread(
    'images/0529/A - 3_fld-182/105/label_map.tif')

plt.subplot(1, 3, 1)
plt.imshow(original_image_8bit)
plt.title('Original')
plt.subplot(1, 3, 2)
plt.title('C3 (log)')
plt.imshow(c3_log_image, cmap='jet')
plt.subplot(1, 3, 3)
plt.imshow(label_map, cmap='jet')
plt.title('Label Map')
plt.tight_layout()
plt.savefig("output/figures/0529_A - 3_fld-182_105.svg", dpi=300)
```
